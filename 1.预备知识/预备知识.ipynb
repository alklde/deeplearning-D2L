## 4. 微积分

```python
# Derivatives visualization: f(x) = 3x^2 - 4x
import numpy as np
def f(x):
    return 3 * x ** 2 - 4 * x

def numerical_lim(f, x, h):
    return (f(x + h) - f(x)) / h

h = 0.1
for i in range(5):
    print(f'h={h:.5f}, numerical limit={numerical_lim(f, 1, h):.5f}')
    h *= 0.1
```

```python
# Visualize f(x) and its tangent at x=1
x_vals = np.arange(-2, 2, 0.1)
y_vals = f(x_vals)
plt.plot(x_vals, y_vals, label='f(x) = 3x^2 - 4x')
plt.scatter(1, f(1), color='red', zorder=5)
tangent_slope = numerical_lim(f, 1, 0.00001)
tangent_line = tangent_slope * (x_vals - 1) + f(1)
plt.plot(x_vals, tangent_line, '--', label=f'Tangent (slope={tangent_slope:.1f})')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Partial derivatives
def g(x1, x2):
    return x1 ** 2 + 2 * x2 ** 2

h = 0.001
x1, x2 = 1.0, 2.0
dg_dx1 = (g(x1 + h, x2) - g(x1, x2)) / h
dg_dx2 = (g(x1, x2 + h) - g(x1, x2)) / h
dg_dx1, dg_dx2
```

## 5. 自动微分

```python
# Simple gradient: y = 2x^T x
x = torch.arange(4.0, requires_grad=True)
y = 2 * torch.dot(x, x)
y.backward()
x.grad, x.grad == 4 * x
```

```python
# Backward for non-scalar functions
x = torch.arange(4.0, requires_grad=True)
y = x * x
y.sum().backward()
x.grad
```

```python
# Detach
x = torch.arange(4.0, requires_grad=True)
y = x * x
u = y.detach()
z = u * x
z.sum().backward()
x.grad, u
```

```python
# Computing gradients of non-scalar with backward
x = torch.arange(4.0, requires_grad=True)
y = x * x
y.backward(torch.ones(len(x)))
x.grad
```

```python
# Gradient accumulation: zero_grad
x = torch.arange(4.0, requires_grad=True)
y = x * x
y.sum().backward()
x.grad
# Without zeroing, gradients accumulate
y = x * x
y.sum().backward()
x.grad
```

```python
# Zero grad properly
x.grad.zero_()
y = x * x
y.sum().backward()
x.grad
```

## 6. 概率

```python
# Basic probability: dice rolling
fair_probs = torch.ones(6) / 6
fair_probs
```

```python
# Sampling from a multinomial distribution
counts = torch.multinomial(fair_probs, 1000, replacement=True)
counts[:20], counts.shape
```

```python
# Empirical frequencies
counts = torch.multinomial(fair_probs, 1000, replacement=True)
relative_freqs = counts.float().bincount(minlength=6) / 1000
relative_freqs
```

```python
# torch.distributions
from torch.distributions import normal, bernoulli
m = normal.Normal(0, 1)
m.sample((5,))
```

```python
# Bernoulli distribution
b = bernoulli.Bernoulli(0.5)
b.sample((10,))
```

```python
# Visualizing probability distribution
counts = torch.multinomial(fair_probs, 10000, replacement=True)
freqs = counts.float().bincount(minlength=6) / 10000
plt.bar(range(1, 7), freqs.numpy(), alpha=0.7, label='Empirical')
plt.axhline(1/6, color='r', linestyle='--', label='Theoretical (1/6)')
plt.xlabel('Face')
plt.ylabel('Probability')
plt.legend()
plt.show()
```
